```python
import json
```

```python
with open("data/mcp_tools_with_embedding.json") as f:
    data = json.load(f)
```

```python
# remove embeddings from both top level and tools
for item in data:
    # Remove top-level embeddings
    if "description_embedding" in item:
        del item["description_embedding"]
    if "summary_embedding" in item:
        del item["summary_embedding"]
    # Remove description_embedding from each tool
    for tool in item.get("tools", []):
        if "description_embedding" in tool:
            del tool["description_embedding"]
        if "summary_embedding" in tool: 
            del tool["summary_embedding"]
```
